```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
df=pd.read_csv(r"C:\Users\vic\AppData\Local\Packages\5319275A.WhatsAppDesktop_cv1g1gvanyjgm\LocalState\sessions\79EB2838B5E78A914D72B12C2CA9A251AEF57AEA\transfers\2026-24\train.csv")
df
pd.set_option('display.max_columns',None)
df
pd.set_option('display.max_rows',None)
df
#Check null values
df.isnull().sum()
df_null_clm=df.isnull().sum()/df.shape[0]*100
df_null_clm
# check null values greater than 20
df_null_clm_list=df_null_clm[df_null_clm>20]
df_null_clm_list
# drop columns greater than 20
df1=df.drop(columns=df_null_clm_list.index)
df1
df1.isnull().sum()/df1.shape[0]*100
#numeric values
df_num_clm=df1.select_dtypes(include=['int','float'])
df_num_clm
#cat values
df_cat_clm=df1.select_dtypes(include=['object'])
df_cat_clm
#null values in numeric col
df_num_clm.isnull().sum()/df_num_clm.shape[0]*100
#null values in cat col.
df_cat_clm.isnull().sum()/df_cat_clm.shape[0]*100
df_num_clm.columns
df_cat_clm.columns
# replace null values in num col
df_num_clm.isnull().sum()/df_num_clm.shape[0]*100 
df_num_clm = df_num_clm.fillna(df_num_clm.median())
df_num_clm.isnull().sum().sum()
# replace null values in cat col
df_cat_clm.isnull().sum()/df_cat_clm.shape[0]*100 
df_cat_clm = df_cat_clm.fillna(df_cat_clm.loc[0])
df_cat_clm.isnull().sum().sum()
# concat both the columns
df_new=pd.concat([df_num_clm,df_cat_clm],axis=1)
df_new
df_cat_clm

#-------------Encoding------------
quality_map = {                                   
    'Po': 1,
    'Fa': 2,
    'TA': 3,
    'Gd': 4,
    'Ex': 5
}

ordinal_cols = [
    'ExterQual',
    'ExterCond',
    'KitchenQual',
    'GarageQual',
    'GarageCond'
]

for col in ordinal_cols:
    df_new[col] = df_new[col].map(quality_map)
df_new[col]

df_new[ordinal_cols].head()# Check result
remaining_cat_cols = df_new.select_dtypes(include='object').columns  # one hot encoding ramining column
df_new = pd.get_dummies(df_new,columns=df_new.select_dtypes(include='object').columns,drop_first=True,dtype='int')

df_new
X = df_new.drop('SalePrice', axis=1)
y = df_new['SalePrice']

from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.2,random_state=12)

from sklearn.preprocessing import StandardScaler

sc = StandardScaler()
cols = X.columns
X_train = sc.fit_transform(X_train)
X_test = sc.transform(X_test)

X_Train_scale = pd.DataFrame(X_train, columns=cols)
X_Test_scale = pd.DataFrame(X_test, columns=cols)
X_Train_scale.head()

from sklearn.linear_model import LinearRegression
lr=LinearRegression()
lr.fit(X_train,y_train)
y_pred=lr.predict(X_test)
lr.score(X_test,y_test)

from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

mean_absolute_error(y_test, y_pred)
mean_squared_error(y_test, y_pred)
mean_squared_error(y_test, y_pred)**0.5
r2_score(y_test, y_pred)

```
